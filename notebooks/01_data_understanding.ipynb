```python
import pandas as pd

data_path = "../data/20261010091500.gkg.csv"

df = pd.read_csv(
    data_path,
    sep="\t",
    header=None,
    low_memory=False
)

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# Display all column names
print("Column Names:\n")

for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")

# Display first 5 rows
print("\nFirst 5 rows:")
display(df.head())
```

```python
columns = [
    "GKGRECORDID", "V2.1DATE", "SourceCollectionIdentifier",
    "SourceCommonName", "DocumentIdentifier", "Counts",
    "V2Counts", "Themes", "V2Themes", "Locations",
    "V2Locations", "Persons", "V2Persons", "Organizations",
    "V2Organizations", "V2Tone", "Dates", "GCAM",
    "SharingImage", "RelatedImages", "SocialImageEmbeds",
    "SocialVideoEmbeds", "Quotations", "AllNames",
    "Amounts", "TranslationInfo", "Extras"
]

df.columns = columns

print("Columns updated:", len(df.columns))
display(df[["SourceCommonName", "DocumentIdentifier", "Themes",
            "Locations", "Persons", "Organizations"]].head(5))
```

```python
# 1. Dataset size
print("Dataset shape:", df.shape)

# 2. Missing values
print("\nMissing values:")
display(
    df.isnull().sum()
       .sort_values(ascending=False)
       .to_frame("Missing Count")
       .head(10)
)

# 3. Duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# 4. Unique news sources
print("\nUnique news sources:", df["SourceCommonName"].nunique())

# 5. Unique news articles
print("\nUnique articles:", df["DocumentIdentifier"].nunique())
```

```python
df["Organization"] = df["Organizations"].str.split(";")

df[["DocumentIdentifier", "Organization"]].head(3)
```

```python
org_df = df[["DocumentIdentifier", "Organization"]].explode("Organization")

org_df.head(5)
```

```python
org_df = org_df.dropna(subset=["Organization"])
org_df["Organization"] = org_df["Organization"].str.strip()

org_df.head(5)
```

```python
org_df["Organization"] = org_df["Organization"].str.lower().str.strip()
```

```python
org_df["Organization"].nunique()
```

```python
import networkx as nx

G = nx.Graph()

for _, row in org_df.iterrows():
    G.add_edge(row["DocumentIdentifier"], row["Organization"])

print("Nodes:", G.number_of_nodes())
print("Edges:", G.number_of_edges())
```

```python
sorted(G.degree, key=lambda x: x[1], reverse=True)[:10]
```

```python
org_df["Organization"].value_counts().head(10)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))
nx.draw(G, node_size=10, with_labels=False)
plt.show()
```

```python
print("Nodes:", G.number_of_nodes())
print("Edges:", G.number_of_edges())
print("Density:", nx.density(G))
```

```python
nx.number_connected_components(G)
```

```python
max(len(c) for c in nx.connected_components(G))
```

```python
largest = max(nx.connected_components(G), key=len)
sum(1 for node in largest if node in set(org_df["Organization"]))
```

```python
org_df[org_df["Organization"].isin(largest)].groupby("Organization").size().sort_values(ascending=False).head(10)
```

```python
subgraph = G.subgraph(largest)
plt.figure(figsize=(12, 8))
nx.draw(subgraph, node_size=10, with_labels=False)
plt.show()
```

```python
top_orgs = org_df["Organization"].value_counts().head(10).index

G_top = G.subgraph(
    [n for n in G.nodes if n in top_orgs or n in org_df["DocumentIdentifier"].values]
)
```

```python
plt.figure(figsize=(12, 8))
pos = nx.spring_layout(G_top, seed=42)
nx.draw(G_top, pos, node_size=15, with_labels=False)
plt.show()
```

```python
top_orgs = org_df["Organization"].value_counts().head(10).index

top_edges = org_df[org_df["Organization"].isin(top_orgs)]

G_top = nx.Graph()

for _, row in top_edges.iterrows():
    G_top.add_edge(row["DocumentIdentifier"], row["Organization"])
```

```python
nx.draw_networkx_labels(
    G_top,
    pos,
    labels={n: n for n in top_orgs if n in G_top.nodes},
    font_size=8
)
plt.show()
```

```python
plt.figure(figsize=(14, 8))

pos = nx.spring_layout(G_top, seed=42)

nx.draw_networkx_edges(G_top, pos, alpha=0.3)
nx.draw_networkx_nodes(G_top, pos, node_size=30)

labels = {n: n for n in top_orgs if n in G_top.nodes}

nx.draw_networkx_labels(G_top, pos, labels=labels, font_size=8)

plt.axis("off")
plt.show()
```

```python
print("Nodes:", G_top.number_of_nodes())
print("Edges:", G_top.number_of_edges())
print("Connected groups:", nx.number_connected_components(G_top))
```

```python
components = sorted(nx.connected_components(G_top), key=len, reverse=True)

print("Largest group:", len(components[0]))
print("Second group:", len(components[1]))
```

```python
df.groupby("SourceCommonName").agg(
    article_count=("DocumentIdentifier", "count"),
    average_tone=("Tone", "mean")
).sort_values("article_count", ascending=False).head(10)
```

```python
source_stats = df.groupby("SourceCommonName").agg(
    article_count=("DocumentIdentifier", "count"),
    average_tone=("Tone", "mean")
).sort_values("article_count", ascending=False).head(10)

source_stats["article_count"].plot(
    kind="bar",
    title="Top 10 News Sources by Article Count"
)

plt.ylabel("Number of Articles")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
source_stats["average_tone"].sort_values().plot(
    kind="bar",
    title="Average News Tone by Source"
)

plt.axhline(0, linestyle="--")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
source_stats["average_tone"].sort_values().plot(
    kind="barh",
    title="Average News Tone by Source"
)

plt.axvline(0, linestyle="--")
plt.tight_layout()
plt.show()
```

```python
source_stats["average_tone"].sort_values(ascending=False).plot(
    kind="barh",
    title="Average News Tone by Source"
)

plt.axvline(0, linestyle="--")
plt.tight_layout()
plt.show()
```
